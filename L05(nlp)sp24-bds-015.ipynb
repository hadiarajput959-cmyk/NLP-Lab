# Task 2: Generating N-grams

```python
def get_ngrams(tokens, n):
    if n <= 0:
        raise ValueError("n must be positive")

    return [
        tuple(tokens[i:i + n])
        for i in range(len(tokens) - n + 1)
    ]


first_sentence = train_tokens[0]

for n in (1, 2, 3):
    print(f"\n{n}-grams:")
    print(get_ngrams(first_sentence, n))


print("\nVerifying N-gram Counts:")
for tokens in train_tokens + test_tokens:
    m = len(tokens)

    for n in (1, 2, 3):
        ngrams = get_ngrams(tokens, n)
        expected = max(0, m - n + 1)

        assert len(ngrams) == expected
        print(f"m={m}, n={n}, actual={len(ngrams)}, expected={expected}")


print("\nN larger than sentence length:")
print(get_ngrams(first_sentence, len(first_sentence) + 1))
```

# Task 3: Counting N-grams

```python
from collections import Counter

unigram_counts = Counter()
bigram_counts = Counter()
trigram_counts = Counter()

for tokens in train_tokens:
    unigram_counts.update(get_ngrams(tokens, 1))
    bigram_counts.update(get_ngrams(tokens, 2))
    trigram_counts.update(get_ngrams(tokens, 3))


def print_counts(title, counts):
    print(f"\n{title}")
    for ngram, count in sorted(
        counts.items(),
        key=lambda item: (-item[1], item[0])
    ):
        print(ngram, ":", count)


print_counts("Unigram Counts", unigram_counts)
print_counts("Bigram Counts", bigram_counts)
print_counts("Trigram Counts", trigram_counts)

print("\nFive Most Common Bigrams:")
for bigram, count in bigram_counts.most_common(5):
    print(bigram, ":", count)

print("\nDistinct Unigrams:", len(unigram_counts))
print("Distinct Bigrams:", len(bigram_counts))
print("Distinct Trigrams:", len(trigram_counts))


def count_of(ngram):
    n = len(ngram)

    if n == 1:
        return unigram_counts.get(tuple(ngram), 0)
    elif n == 2:
        return bigram_counts.get(tuple(ngram), 0)
    elif n == 3:
        return trigram_counts.get(tuple(ngram), 0)

    return 0


print("\nSeen bigram:", count_of(("the", "cat")))
print("Unseen bigram:", count_of(("fish", "dog")))
```

# Task 4: Unigram Model

```python
import math
from itertools import permutations

# Exclude <s> because it is not a predicted word.
unigram_model_counts = Counter({
    (word,): count
    for (word,), count in unigram_counts.items()
    if word != "<s>"
})

N = sum(unigram_model_counts.values())


def unigram_prob(w):
    return unigram_model_counts.get((w,), 0) / N


print("Word\t\tProbability")
for word in sorted(V):
    print(f"{word:<12}{unigram_prob(word):.4f}")

assert math.isclose(
    sum(unigram_prob(w) for w in V), 1.0
)

print("\nUnigram Sentence Probabilities:")


def unigram_sentence_prob(sentence):
    tokens = tokenize(sentence)
    return math.prod(
        unigram_prob(w)
        for w in tokens[1:]
    )


for sentence in test:
    print(sentence, ":", f"{unigram_sentence_prob(sentence):.4f}")


print("\nDifferent Word Orders:")
words = test[0].split()

for order in list(permutations(words))[:6]:
    sentence = " ".join(order)
    print(sentence, ":", f"{unigram_sentence_prob(sentence):.8f}")
```

# Task 5: Bigram Model (MLE)

```python
from collections import defaultdict

context_counts = Counter(
    prev for (prev, word), count in bigram_counts.items()
    for _ in range(count)
)


def bigram_prob(prev, w):
    denominator = context_counts.get(prev, 0)

    if denominator == 0:
        return 0.0

    return bigram_counts.get((prev, w), 0) / denominator


def next_word_distribution(prev):
    denominator = context_counts.get(prev, 0)

    if denominator == 0:
        return {}

    return {
        word: bigram_prob(prev, word)
        for word in sorted(V)
        if bigram_prob(prev, word) > 0
    }


print("Next-word Distribution for 'the':")
distribution = next_word_distribution("the")

for word, prob in distribution.items():
    print(f"{word:<12}{prob:.4f}")

assert math.isclose(sum(distribution.values()), 1.0)


def bigram_sentence_prob(sentence):
    tokens = tokenize(sentence)

    return math.prod(
        bigram_prob(tokens[i - 1], tokens[i])
        for i in range(1, len(tokens))
    )


print("\nBigram Sentence Probabilities:")
for sentence in test:
    print(sentence, ":", f"{bigram_sentence_prob(sentence):.8f}")


def predict_next(prev):
    distribution = next_word_distribution(prev)

    if not distribution:
        return None

    return min(
        distribution,
        key=lambda word: (-distribution[word], word)
    )


print("\nNext-word Predictions:")
for word in ["the", "cat", "dog", "sat", "ate", "fish"]:
    print(word, "->", predict_next(word))
```

# Task 6: Trigram Model

```python
trigram_context_counts = Counter()

for (w1, w2, w3), count in trigram_counts.items():
    trigram_context_counts[(w1, w2)] += count


def trigram_prob(w1, w2, w3):
    denominator = trigram_context_counts.get((w1, w2), 0)

    if denominator == 0:
        return 0.0

    return trigram_counts.get((w1, w2, w3), 0) / denominator


def trigram_sentence_prob(sentence):
    tokens = tokenize(sentence)

    if len(tokens) < 2:
        return 1.0

    probability = bigram_prob(tokens[0], tokens[1])

    for i in range(2, len(tokens)):
        probability *= trigram_prob(
            tokens[i - 2], tokens[i - 1], tokens[i]
        )

    return probability


print("Trigram Sentence Probabilities:")
for sentence in test:
    print(sentence, ":", f"{trigram_sentence_prob(sentence):.8f}")


print("\nDistinct N-gram Counts")
print(f"{'N-gram':<12}{'Distinct Count'}")
print(f"{'Unigrams':<12}{len(unigram_counts)}")
print(f"{'Bigrams':<12}{len(bigram_counts)}")
print(f"{'Trigrams':<12}{len(trigram_counts)}")


for sentence in test:
    tokens = tokenize(sentence)
    unseen_bi = [
        ng for ng in get_ngrams(tokens, 2)
        if bigram_counts.get(ng, 0) == 0
    ]
    unseen_tri = [
        ng for ng in get_ngrams(tokens, 3)
        if trigram_counts.get(ng, 0) == 0
    ]

    print(f"\nSentence: {sentence}")
    print("Unseen bigrams:", unseen_bi)
    print("Unseen trigrams:", unseen_tri)
    print(
        "Model with more unseen N-grams:",
        "Trigram" if len(unseen_tri) > len(unseen_bi)
        else "Bigram" if len(unseen_bi) > len(unseen_tri)
        else "Equal"
    )
```

# Task 8: Smoothing

```python
def bigram_prob_addone(prev, w):
    numerator = bigram_counts.get((prev, w), 0) + 1
    denominator = context_counts.get(prev, 0) + len(V)

    return numerator / denominator


print("MLE vs Add-one Smoothing")
print(f"{'Probability':<22}{'MLE':<12}{'Add-one'}")

comparisons = [
    ("P(sat | fish)", "fish", "sat"),
    ("P(cat | the)", "the", "cat")
]

for label, prev, word in comparisons:
    print(
        f"{label:<22}"
        f"{bigram_prob(prev, word):<12.4f}"
        f"{bigram_prob_addone(prev, word):.4f}"
    )


# Verify that probabilities sum to 1 for each context.
for prev in V | {"<s>"}:
    total = sum(
        bigram_prob_addone(prev, word)
        for word in V
    )
    assert math.isclose(total, 1.0)

print("\nAll add-one probability distributions sum to 1.")


def bigram_prob_addk(prev, w, k):
    if k <= 0:
        raise ValueError("k must be positive")

    numerator = bigram_counts.get((prev, w), 0) + k
    denominator = context_counts.get(prev, 0) + k * len(V)

    return numerator / denominator


print("\nP(sat | fish) with Different k Values")
print(f"{'k':<12}{'Probability'}")

for k in [1, 0.5, 0.1, 0.01]:
    print(f"{k:<12}{bigram_prob_addk('fish', 'sat', k):.4f}")
```

# Task 9: Perplexity and Model Comparison

```python
def perplexity(sentences, prob_fn):
    total_log_prob = 0.0
    total_tokens = 0

    for sentence in sentences:
        probability = prob_fn(sentence)

        if probability <= 0:
            return float("inf")

        total_log_prob += math.log(probability)
        total_tokens += len(tokenize(sentence)) - 1

    if total_tokens == 0:
        return float("inf")

    return math.exp(-total_log_prob / total_tokens)


def unigram_prob_addone(w):
    return (
        unigram_model_counts.get((w,), 0) + 1
    ) / (N + len(V))


def unigram_sentence_prob_addone(sentence):
    tokens = tokenize(sentence)

    return math.prod(
        unigram_prob_addone(w)
        for w in tokens[1:]
    )


def make_addk_sentence_prob(k):
    def sentence_probability(sentence):
        tokens = tokenize(sentence)

        return math.prod(
            bigram_prob_addk(tokens[i - 1], tokens[i], k)
            for i in range(1, len(tokens))
        )

    return sentence_probability


print("Perplexity Comparison on Test Set")
print(f"{'Model':<25}{'Perplexity'}")

models = {
    "Bigram MLE": bigram_sentence_prob,
    "Bigram Add-one": make_addk_sentence_prob(1),
    "Unigram Add-one": unigram_sentence_prob_addone
}

for name, prob_fn in models.items():
    pp = perplexity(test, prob_fn)
    print(f"{name:<25}{pp:.4f}")


print("\nAdd-k Bigram Perplexity")
print(f"{'k':<12}{'Perplexity'}")

k_values = [1, 0.5, 0.1, 0.01]
test_k_results = {}

for k in k_values:
    pp = perplexity(test, make_addk_sentence_prob(k))
    test_k_results[k] = pp
    print(f"{k:<12}{pp:.4f}")

best_test_k = min(test_k_results, key=test_k_results.get)

print("Best k directly on test set:", best_test_k)
print("Best test-set perplexity:", f"{test_k_results[best_test_k]:.4f}")


# Train on the first three sentences.
# Keep V built from all four training sentences.
train_subset = train[:3]
validation = [train[3]]

subset_tokens = [tokenize(s) for s in train_subset]

subset_unigram_counts = Counter()
subset_bigram_counts = Counter()
subset_trigram_counts = Counter()

for tokens in subset_tokens:
    subset_unigram_counts.update(get_ngrams(tokens, 1))
    subset_bigram_counts.update(get_ngrams(tokens, 2))
    subset_trigram_counts.update(get_ngrams(tokens, 3))


# Save the original counts before validation training.
original_unigram_counts = unigram_counts
original_bigram_counts = bigram_counts
original_trigram_counts = trigram_counts
original_context_counts = context_counts
original_unigram_model_counts = unigram_model_counts
original_N = N

# Replace counts with the first three sentences.
unigram_counts = subset_unigram_counts
bigram_counts = subset_bigram_counts
trigram_counts = subset_trigram_counts

context_counts = Counter()
for (prev, word), count in bigram_counts.items():
    context_counts[prev] += count

unigram_model_counts = Counter({
    (word,): count
    for (word,), count in unigram_counts.items()
    if word != "<s>"
})

N = sum(unigram_model_counts.values())


# Choose k using the fourth sentence as validation data.
validation_results = {}

for k in k_values:
    validation_results[k] = perplexity(
        validation,
        make_addk_sentence_prob(k)
    )

best_validation_k = min(
    validation_results,
    key=validation_results.get
)

print("\nValidation Perplexity (First 3 Sentences Used for Training)")
print(f"{'k':<12}{'Perplexity'}")

for k, pp in validation_results.items():
    print(f"{k:<12}{pp:.4f}")

print("Best k on validation set:", best_validation_k)


# Evaluate the selected k on the test set.
validation_selected_test_pp = perplexity(
    test,
    make_addk_sentence_prob(best_validation_k)
)

print(
    "\nTest perplexity using validation-selected k:",
    f"{validation_selected_test_pp:.4f}"
)

print("Best k selected directly on test set:", best_test_k)
print(
    "Best test-set perplexity:",
    f"{test_k_results[best_test_k]:.4f}"
)


# Restore the original full-training counts.
unigram_counts = original_unigram_counts
bigram_counts = original_bigram_counts
trigram_counts = original_trigram_counts
context_counts = original_context_counts
unigram_model_counts = original_unigram_model_counts
N = original_N
```
