<h1>Task 1 — Regex Fundamentals</h1>

```python
import re
```

```python
text = """
My name is Rida and I am studying NLP in 2026.
I have 3 assignments and 25 exercises.
Python, Regex, and Jupyter are useful tools.
The course code is NLP2026 and my marks are 95.
"""
```

**a) Extract all digits**

```python
digits = re.findall(r'\d', text)

print("Digits:")
print(digits)
```

**b) Extract all words that start with a capital letter**

```python
capital_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)

print("Words starting with a capital letter:")
print(capital_words)
```

**c) Extract all sequences of exactly 4 digits**

```python
four_digit_numbers = re.findall(r'\b\d{4}\b', text)

print("Exactly 4-digit sequences:")
print(four_digit_numbers)
```

**d) Extract words between 3 to 6 characters long**

```python
words_3_to_6 = re.findall(r'\b[A-Za-z]{3,6}\b', text)

print("Words with 3 to 6 characters:")
print(words_3_to_6)
```

<h1>Task 2 — Pattern-Based Extraction</h1>

```python
import re
paragraph = """
Please contact us at support@example.com or admin@gmail.com.
For assistance, call 0300-1234567 or 0312-7654321.
Order IDs are ORD12345 and ORD67890.
The order was placed on 25/09/2026 and delivered on 27/09/2026.
The discount offered was 45% and another customer received 12.5% discount.
"""
```

**a) Extract all email addresses**

```python
emails = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', paragraph)

print("Email addresses:")
print(emails)
```

**b) Extract all phone numbers**

```python
phone_numbers = re.findall(r'\b03\d{2}-\d{7}\b', paragraph)

print("Phone numbers:")
print(phone_numbers)
```

**c) Extract all dates in DD/MM/YYYY format**

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', paragraph)

print("Dates:")
print(dates)
```

**d) Extract all percentage values**

```python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', paragraph)

print("Percentage values:")
print(percentages)
```

```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
paragraph = """
I don't like wasting time when I am studying NLP.
It's important to practice every day because practice improves understanding.
Python makes NLP tasks easier and more interesting.
I can't wait to build my own NLP project.
"""
```

**a) Tokenize using NLTK word_tokenize()**

```python
nltk_tokens = word_tokenize(paragraph)

print("NLTK Tokens:")
print(nltk_tokens)
```

**b) Tokenize using a blank spaCy pipeline**

```python
nlp = spacy.blank("en")

doc = nlp(paragraph)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)
```

# Comparison:
# 1. NLTK word_tokenize() produced 44 tokens, while spaCy's blank
#    English pipeline produced 49 tokens.
#
# 2. The difference occurs because NLTK and spaCy use different
#    tokenization rules, especially for contractions and punctuation.
#
# 3. For example, spacy count /n as token too while nltk ignore it
#
# Therefore, NLTK produced 44 tokens and spaCy produced 49 tokens
# for the same paragraph.

**d) Count total tokens**

```python
nltk_count = len(nltk_tokens)
spacy_count = len(spacy_tokens)

print("Total NLTK tokens:", nltk_count)
print("Total spaCy tokens:", spacy_count)
```

<h1>Task 4 — Token Attributes</h1>

```python
import spacy
nlp = spacy.blank("en")
sentence = "Rida bought 2 books for $50 and contacted shop@example.com."
```

**a) Load sentence into a spaCy Doc object**

```python
doc = nlp(sentence)

print("Tokens:")
for token in doc:
    print(token.text)
```

**b) Print is_alpha**

```python
print("Token\t\tis_alpha")

for token in doc:
    print(f"{token.text}\t\t{token.is_alpha}")
```

**c) Print like_num and like_email**

```python
print("Token\t\tlike_num\tlike_email")

for token in doc:
    print(f"{token.text}\t\t{token.like_num}\t\t{token.like_email}")
```

**d) Identify tokens where is_currency is True**

```python
currency_tokens = []

for token in doc:
    if token.is_currency:
        currency_tokens.append(token.text)

print("Currency tokens:")
print(currency_tokens)
```
