<h1>TASK # 8</h1>

<h3>8a</h3>

```python

def clean(text):
    return text.strip().lower()

clean_lambda = lambda t: t.strip().lower()

sample = "  HELLO Natural Language Processing  "
print("Function:", clean(sample))
print("Lambda:", clean_lambda(sample))
```

<h3>8b</h3>

```python

def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
tokens = tokenize(clean(sentence))
filtered = remove_short(tokens, 3)
print("Filtered tokens:", filtered)
```

<h3>8c</h3>

```python

def word_lengths(tokens):
    return {t: len(t) for t in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print("Word lengths:", word_lengths(tokens))
```

<h3>8d</h3>

```python

def nlp_pipeline(text, min_len):
    text = clean(text)
    tokens = tokenize(text)
    tokens = remove_short(tokens, min_len)
    return word_lengths(tokens)

print(nlp_pipeline("Deep learning powers NLP models", 3))
print(nlp_pipeline("Python is great for text analysis", 4))
```
